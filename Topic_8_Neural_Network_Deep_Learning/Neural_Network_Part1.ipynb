### Applying mathematical operations to tensors

```python
torch.manual_seed(1)

t1 = 2 * torch.rand(5, 2) - 1
t2 = torch.normal(mean=0, std=1, size=(5, 2))
```

```python
t3 = torch.multiply(t1, t2)
print(t3)
```

```python
t4 = torch.mean(t1, axis=0)
print(t4)
```

```python
t5 = torch.matmul(t1, torch.transpose(t2, 0, 1))

print(t5)
```

```python
t6 = torch.matmul(torch.transpose(t1, 0, 1), t2)

print(t6)
```

```python
norm_t1 = torch.linalg.norm(t1, ord=2, dim=1)

print(norm_t1)
```

```python
np.sqrt(np.sum(np.square(t1.numpy()), axis=1))
```

# Automatic differentiation

```python
x = torch.tensor([1.7],dtype=torch.float32)
w = torch.tensor([4.2],dtype=torch.float32, requires_grad=True)

def model(w, x):
  return w * w * x

y = model(w, x)

y.backward()

print(y)
print(w.grad)
```

### Always use `torch.no_grad()` when evaluating a model to avoid corrupting the gradients

```python
new_data = torch.tensor([2.7],dtype=torch.float32)

with torch.no_grad():
  y = model(w, new_data)

print(y)
print(w.grad)
```

## Dataloader in PyTorch

```python
# Load molecular solvation data
import pandas as pd

df = pd.read_csv('solubility_data.csv')

df.head(10)
```

```python


X = df[['Molecular Weight','Polar Surface Area']]
X.head()
```

```python
from torch.utils.data import DataLoader, TensorDataset

# Convert DataFrame to a NumPy array and then to a PyTorch Tensor
tensor = torch.tensor(X.values[:10])

# Create a TensorDataset and DataLoader
dataset = TensorDataset(tensor)
dataloader = DataLoader(dataset, batch_size=5, shuffle=True)

for batch in dataloader:
    print(batch)
```
