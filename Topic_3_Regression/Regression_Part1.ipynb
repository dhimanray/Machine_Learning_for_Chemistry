# Stochastic Gradient Descent Linear Regression with SkLearn

```python
from sklearn.linear_model import SGDRegressor

sgd_reg = SGDRegressor(max_iter=1000, tol=1e-5, penalty=None, eta0=0.01, learning_rate='constant',verbose=5)
sgd_reg.fit(X, y.ravel()) # y.ravel() because fit() expects 1D targets

```

```python
sgd_reg.intercept_, sgd_reg.coef_
```

```python
plt.plot(X, y, "b.")

y_predict = sgd_reg.predict(X)
plt.plot(X,y_predict,"k")
plt.xlabel("$x_1$")
plt.ylabel("$y$", rotation=0)
plt.axis([0, 2, 0, 15])
plt.grid()
plt.show()
```

# Test on Molecular Solvation Data

```python
import pandas as pd

df = pd.read_csv('solubility_data.csv')

df.head(10)
```

```python
import seaborn as sns

data = df[['ESOL predicted log solubility in mols per litre','Minimum Degree','Molecular Weight','Number of H-Bond Donors','Number of Rings','Number of Rotatable Bonds','Polar Surface Area','measured log solubility in mols per litre']]
sns.pairplot(data)
```

```python
X = df[['Molecular Weight','Number of H-Bond Donors','Number of Rings','Number of Rotatable Bonds','Polar Surface Area']].to_numpy()
y = df[['measured log solubility in mols per litre']].to_numpy()
```

```python
from sklearn.metrics import r2_score

#use pipeline to make model
model =  SGDRegressor(max_iter=1000, tol=1e-5, penalty=None, eta0=0.01, learning_rate='constant')
model.fit(X,y.ravel())

#plot output
y_predict = model.predict(X)

plt.scatter(y,y_predict,color='r')
plt.plot(y,y)

#calculate r2 score
r2 = r2_score(y,y_predict)
plt.text(-10,0,'R$^2$ = %0.3f'%r2)


plt.xlabel('Target')
plt.ylabel('Prediction')

```

```python
from sklearn.preprocessing import StandardScaler

#Feature scaling
scaler = StandardScaler()
scaled_X = scaler.fit_transform(X)

model =  SGDRegressor(max_iter=1000, tol=1e-5, penalty=None, eta0=0.01, learning_rate='constant')
model.fit(scaled_X,y.ravel())

#plot output
y_predict = model.predict(scaled_X)

plt.scatter(y,y_predict,color='r')
plt.plot(y,y)

#calculate r2 score
r2 = r2_score(y,y_predict)
plt.text(-10,0,'R$^2$ = %0.3f'%r2)


plt.xlabel('Target')
plt.ylabel('Prediction')
plt.title('weights = '+str(model.coef_)+' bias = '+str(model.intercept_),fontsize=10)
```
